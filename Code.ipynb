```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('data.csv')
df.head()
```

```python
df.columns
```

```python
# getting rid of -1 to make them 0
numeric_cols = ['videoViewCount', 'videoLikeCount', 'videoDislikeCount', 'VideoCommentCount']
for col in numeric_cols:
    df[col] = df[col].replace(-1, 0)
```

```python
# Question 1: Calculate the average number of likes per video.

average_likes = df['videoLikeCount'].mean()
average_likes
```

```python
# Question 2:  Compute the engagement rate for each video and find average engagement rate

engagements = df['videoLikeCount'] + df['videoDislikeCount'] + df['VideoCommentCount']
views = df['videoViewCount']

df["engagement_rate"] = engagements / views
df["engagement_rate"] = df["engagement_rate"].replace([float('inf'), float('nan')], 0)
    
average_engagement_rate = df["engagement_rate"].mean()
round(average_engagement_rate, 4)
```

```python
# Question 3: Calculate the combined storage size for a single video in megabytes (MiB)

metadata_bytes = 500
thumbnail_bytes = 200 * 1024
total_bytes = metadata_bytes + thumbnail_bytes

total_megabytes = total_bytes / (1024 ** 2)
total_megabytes_rounded = round(total_megabytes, 4)
total_megabytes_rounded
```

```python
# Question 4: Estimate total data transfer for all engagements within a dataset

total_engagements = engagements.sum()
total_data_transfer_mib = total_engagements * total_megabytes
total_data_transfer_mib_rounded = round(total_data_transfer_mib, 4)
total_data_transfer_mib_rounded
```

```python
# Question 5: Estimate number of views per minute over 5 years

total_views = views.sum()
total_minutes = 5 * 365 * 24 * 60
views_per_minute = total_views / total_minutes
views_per_minute
```

```python
# Question 6: Data accessed per minute
data_per_minute = views_per_minute * total_megabytes
data_per_minute = data_per_minute
data_per_minute
```
